```python
import numpy as np
```

```python
# 1D Array Operations

arr = np.arange(1, 21)

# a. Statistics
print("Array:", arr)
print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard Deviation:", np.std(arr))

# b. Indices greater than 10
indices = np.where(arr > 10)
print("Indices where elements > 10:", indices)
```

```python
# 2D Array Operations

arr2d = np.arange(1, 17).reshape(4, 4)

print("2D Array:\n", arr2d)

# Transpose
print("Transpose:\n", arr2d.T)

# Row-wise sum
print("Row-wise sum:", np.sum(arr2d, axis=1))

# Column-wise sum
print("Column-wise sum:", np.sum(arr2d, axis=0))
```

```python
# Random Arrays

a = np.random.randint(1, 21, (3, 3))
b = np.random.randint(1, 21, (3, 3))

print("Array A:\n", a)
print("Array B:\n", b)

print("Addition:\n", a + b)
print("Subtraction:\n", a - b)
print("Multiplication:\n", a * b)

print("Dot Product:\n", np.dot(a, b))
```

```python
# Reshape & Slice

arr = np.arange(1, 13)
reshaped = arr.reshape(3, 4)

print("Reshaped Array:\n", reshaped)

# Slice first 2 rows and last 2 columns
sliced = reshaped[:2, -2:]
print("Sliced Array:\n", sliced)
```

```python
# – Pandas

import pandas as pd
```

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [24, 27, 22, 32, 29],
    'Department': ['HR', 'Finance', 'IT', 'Marketing', 'HR'],
    'Salary': [45000, 54000, 50000, 62000, 47000]
}

df = pd.DataFrame(data)

# a. First 5 rows
print(df.head())

# b. Summary stats
print(df[['Age', 'Salary']].describe())

# c. Average salary in HR
avg_hr = df[df['Department'] == 'HR']['Salary'].mean()
print("Average HR Salary:", avg_hr)
```

```python
df['Bonus'] = df['Salary'] * 0.10
print(df)
```

```python
filtered = df[(df['Age'] >= 25) & (df['Age'] <= 30)]
print(filtered)
```

```python
# 4. Group by Department
grouped = df.groupby('Department')['Salary'].mean()
print(grouped)
```

```python
# Sort & Save CSV

sorted_df = df.sort_values(by='Salary')
sorted_df.to_csv('sorted_salary.csv', index=False)

print(sorted_df)
```

```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [10, 15, 25, 30, 50]

plt.plot(x, y)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid()
plt.show()
```

```python
# 2. Bar Graph

students = ['John', 'Jane', 'Alice', 'Bob']
marks = [75, 85, 60, 90]

plt.bar(students, marks)
plt.title("Student Marks")
plt.show()
```

```python
# Pie Chart

regions = ['North America', 'Europe', 'Asia', 'Others']
revenue = [45, 25, 20, 10]

explode = [0.1, 0, 0, 0]

plt.pie(revenue, labels=regions, explode=explode, autopct='%1.1f%%')
plt.title("Revenue Distribution")
plt.show()
```

```python
# Histogram
data = np.random.randint(1, 101, 1000)

plt.hist(data)
plt.title("Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```
